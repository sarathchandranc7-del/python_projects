## Numpy Array Operations:  

### 1. Create a 1D NumPy Array:

```python
import numpy as np
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
print(temperatures_w1)
```

### 2. Inspection and Properties:

```python
print(temperatures_w1.shape)
```

```python
print(temperatures_w1.dtype)
```

```python
print(temperatures_w1.size)
```

### 3. Array Operations:

```python
Fahrenheit=(temperatures_w1*9/5)+35
```

```python
print(Fahrenheit)
```

```python
print(np.max(temperatures_w1))
```

```python
print(np.min(temperatures_w1))
```

```python
print(np.mean(temperatures_w1))
```

### 4. Array Slicing and Indexing:

```python
print(temperatures_w1[:3])
```

```python
print(temperatures_w1[5:])
```

```python
print(temperatures_w1[3:6])
```

## Pandas Series:  

### 1. Creating Pandas Series: 

```python
import pandas as pd
```

```python
marks = pd.Series(
    [95,92,89,85,80],
    index=['Rank 1','Rank 2','Rank 3','Rank 4','Rank 5'])
```

```python
print(marks)
```

### 2. Indexing and Slicing:

```python
print(marks.iloc[0])
```

```python
print(marks.loc[['Rank 1','Rank 2','Rank 3']])
```

```python
print(marks.iloc[2])
```

```python
print(marks[marks>90])
```

### 3. Manipulating Series:

```python
marks["Rank 1"]=100
```

```python
print(marks)
```

```python
marks=marks.drop('Rank 5')
```

```python
print(marks)
```

```python
cgpa = marks/10
print(cgpa)
```

## Pandas DataFrame:

### 1. Creating Pandas DataFrame: 

```python
import pandas as pd
```

```python
transactions = pd.DataFrame({
        'TransactionID':[101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory' :['Electronics', 'Clothing', 'Electronics', 'Furniture', 
                    'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
            'Region':['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 
                      'South', 'North'] ,
            'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400 ]})
```

```python
print(transactions)
```

### 2. Data Exploration:

```python
transactions.info()
```

```python
print(transactions.groupby("Region")["Amount"].mean())
```

### 3. Manipulating the DataFrame:

```python
transactions.loc[transactions["TransactionID"]==102, "Amount"]=165
```

```python
def discount_10(x):
    return x * 10/100    
```

```python
transactions["Discount"]=transactions["Amount"].apply(discount_10)
```

```python
print (transactions)
```

```python
transactions.drop(8,axis=0,inplace=True)
```

```python
print (transactions)
```

```python
transactions.drop(["Discount"],axis=1,inplace=True)
```

```python
print (transactions)
```
