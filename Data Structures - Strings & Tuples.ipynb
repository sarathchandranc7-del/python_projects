# Strings & Tuples
### 1.String Concatenation: 

```
welcome = 'Hello '
```

```
name = input("Enter your Name")
result = welcome + name
```

```
print(result)
```

```
message = ', welcome to Python Programming'
Result = result + message
```

```
print(Result)
```

### 2.String Slicing and Indexing:

```
string = 'Hello Python'
```

```
print(string[-1])
```

```
print(string[0])
```

```
print(string[0:5])
```

```
print(string[-11:])
```

```
print(string[::-1])
```

```
print(string[6:12])
```

### 3. String Methods:

```
strM = 'Python beginner tutorial'
```

```
print(strM.upper())
```

```
print(strM.lower())
```

```
print(strM.capitalize())
print(strM)
```

```
print(strM.count('t'))
```

```
print(strM.replace('Python','Machine Learning'))
```

### Tuples (Creation, Modification and Access) :  

```
t1=(10,20,30)
```

```
t2=(40,50,60)
```

```
print(t1+t2)
```

```
combine = t1+t2
```

```
print (combine)
```

```
print(combine*3)
```

```
print(combine[2])
```

```
print(combine[:3])
```

```
print(combine[-3:])
```
